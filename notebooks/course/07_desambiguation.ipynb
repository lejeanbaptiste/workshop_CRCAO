# 7. Désambiguïsation

## Théorie minimale

Jusqu'ici, `persName` décrit une forme dans le texte. Il ne dit pas encore quelle personne cette forme désigne. Une personne peut avoir plusieurs noms ou formes de surface ; plusieurs personnes peuvent porter le même nom. L'ordinateur a donc besoin d'un identifiant stable, par exemple `person-76296`.

Les identifiants d'autorité comme ORCID, DOI ou ISBN sont utiles, mais aucune base n'est complète et chaque projet rencontre des entités absentes, des variantes et des homonymes. La bonne pratique consiste à conserver la provenance de l'identifiant et à documenter les décisions.

## Dans l'éditeur

Le fichier `entities.xml` joue le rôle d'une petite base relationnelle. Il relie un identifiant à des formes de surface, des types, des autorités et des informations utiles au filtrage. Dans `Disambiguate`, examinez les propositions, leurs concordances et leur contexte. Créez une nouvelle entité lorsque rien ne convient ; ne forcez pas une correspondance pour supprimer une absence.

Une occurrence peut alors ressembler à ceci :

```xml
<persName key="person-76296">Jeremy</persName>
```

L'attribut `key` relie la mention dans le texte à l'entité dans la base. D'autres attributs comme `type`, `cert` ou `ref` peuvent être utiles, mais seulement si votre question de recherche justifie leur emploi.

## Exercices

- [ ] Sélectionner une balise et vérifier si elle possède déjà un `key`.
- [ ] Lancer une recherche dans `Disambiguate`.
- [ ] Comparer deux homonymes à partir de leur contexte.
- [ ] Créer ou modifier une entité et lui ajouter des formes de surface.
- [ ] Consulter l'entrée dans `Entities` et noter la source de votre décision.

La désambiguïsation est une interprétation savante. Elle peut être révisée : un identifiant stable rend cette révision explicite au lieu de la cacher dans une correction silencieuse du texte.
