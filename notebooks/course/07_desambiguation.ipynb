# 7. Désambiguïsation

## Pratique traditionnelle

Vous avez oublié les dates de 劉歆, le fameux lettré, bibliothécaire et propagandiste à la fin des Han orrientaux. Vous tapez donc « Liu Xin » dans Google, mais il ne vous propose qu'une présentatrice et une rappeuse. Vous le cherchez ensuite dans Wikipédia, qui vous dirige vers une liste de tous les personnages fameux qui portent le même nom. Ceci s'appelle une « Disambiguation page » :

![Page de désambiguïsation Wikipédia pour « Liu Xin »](images/wikipedia-disambiguation-liu-xin.png)

Comment « disambiguïser » ? Après chaque « Liu Xin » il y a une description courte (« scholar ») et longue « (劉歆; c. 50 BCE – 23 CE), astronomer, historian, librarian, mathematician, politician and Confucian scholar of the Han dynasty ».

Une fois sur la page du bon « Liu Xin », on a ses dates, son nom de style, les membres de sa famille, etc.

## Petite modification

Après avoir balisé votre texte, vous allez lancer le panneau `Disambiguate`. Pour chaque chaîne de caractères balisée, l'éditeur va vous proposer toutes les **« entités »** (personne, lieu, etc.) qui y sont associées dans les bases de données d'autorité. Celles-ci s'appelle des **« authorités »**, comme le [Buddhist Studies Authority Database Project](https://authority.dila.edu.tw/). 

Comme vous le faîtes déjà, vous allez cliquez sur le bon « Liu Xin » dans Wikipédie, et vous allez associer toutes les occurences de ce nom dans votre texte à cette seule et même personne. La différence : 

- vous avez ajouté cette personne dans votre base de données locale ;
- vous lui avez attribué un identifiant unique ;
- vous avez associé cet identifiant unique à celui de l'autorité de manière pérenne et, donc, toutes les données dans cette base de données externes ;
- vous trouverez desormais le lien vers sa page Wikipédia dans le panneau « Attributes », pour votre consultation rapide ;
- vous pouvez le retrouver dans le panneau « Entities », y ajouter des noms alternatifs, défaire son association Wikipédia, etc.

Parfois les autorités se réfèrent l'une à l'autre, et en cliquant sur la bonne personne dans Wikipédia on fait l'association aussi avec un personnage dans CBDB, DILA, VIAF et la NDL. Parfois il y a des doublons idiots à regrouper, si vous le souhaitez ; et parfois il n'y a rien du tout. Dans ce cas-ci, vous pouvez quand même ajouter la chose en question dans votre propre base de données en tant que nouvelle entité. 

Et si l'on tombe sur le prénom 歆, tout seule ? À l'heure actuelle, ce sera ni balisé automatiquement ni recconu par la fonction « Disambiguation ». Ne n'est pas grave : 

- Sélectionner une occurence, 
- Balisez-la comme `persName`
- Ouvrez le panneau « Attributes »
- Cherchez le nom complet dans « Lookup »
- Associez-le à une entité
- Appliquez le balisage et l'association à toutes les autres occurences dans le document.
- Ajouter « 歆 » comme prénom de 劉歆 dans « Entities »

## Attribus

Encore une fois : on distingue entre _les mots_ et _les choses_ – les « chaînes de caractères » et les « entités » —, et on substitue _les choses_ par _les identifiants uniques et universels_. 

L'identifiant est inséré dans la balise en tant qu'**attribu** :

```xml
<persName key="person-000020" resp="#ljb-autotag">許芝</persName>
```

Dans ce cas-ci, `key` indique l'identifiant dan ma base de données, et `resp` l'origine de la balise. Les mêmes informations sont présentées en mode « Visual » dans le panneau « Attributes ». Nous avons parlé à tout à l'heure d'indiquer la certitude d'une identification avec l'attribu `cert`. C'est le même principe. 

- [ ] Ajoutez l'attribu et la valeur `cert="5"` à deux balises.

Pareil, on peut utilisez XPath pour naviguer avec des attribus. Par exemple, si vous voulez trouvez toutes les balises (*) avec l'attribut `cert` :

```xpath
TEI//*[@cert]
```

Toutes les balises avec `cert="5` :

```xpath
TEI//*[@cert="5"]
```

Toutes les `persName` _sans identifiant_ :

```xpath
TEI//persName[not(@key)]
```

Toutes les références, peu importe ce qui est écrit, à la personne `000020` :

```xpath
TEI//persName[@key="person-000020"]
```
