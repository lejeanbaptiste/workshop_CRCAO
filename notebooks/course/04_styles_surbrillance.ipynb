# 4. Styles et surbrillance utile

## Pratique traditionnelle

Vous avez chacun vos habitudes à propos de la surbrillance. Lorsque je lis un article, je met les faits intéressants en jaune, les personnes et les ouvrages en orange, les éléments structurels et argumentaires en vert, et les bêtises en blue. L'objectif est de me repérer et de _baliser_ le texte pour une rélecture rapide.  

Le balisage XML, c'est mettre de la surbrillance dans vos sources primaires, mais de manière précise, systématique et bien ciblée.  

## Petite modification

1. Repérez un nom de personne dans votre texte.
2. Sélectionnez-le et appuyez sur `Entrée`.
3. Choisissez `persName`.
4. Dans le panneau `Attributes`, à droite, changez la couleur de la balise et du texte.
5. Faites la même chose pour deux autres personnes.

Ceci est _très utile_ pour vous repérer dans un texte, en particulier lorsque l'on met les personnes, les lieux, les organisations et les dates en différents couleurs. Mais ce n'est qu'une dérive de ce que vous venez de faire.

**Vous avez fait du balisage**. Dans le code, vous avez repéré une [chaîne de caractères](https://fr.wikipedia.org/wiki/Cha%C3%AEne_de_caract%C3%A8res) – `Daniel` – et vous l'avez remplacée par `<persName>Daniel</persName>`. Le même pour `Jeremy`, etc. Cliquez sur `Show tags` (ICON) et vous voyerez bien. Vous avez non seulement indiqué que les deux s'agissent de noms de personne, mais qu'ils appartient à la même catégorie. 

Dans tout site-web et traitement de texte, le rendu graphique du texte est géré par une CSS ([Cascading Style Sheet](https://fr.wikipedia.org/wiki/Feuilles_de_style_en_cascade)), qui déclare par exemple que tous les paragraphes doivent commencer sur une nouvelle ligne, avec indentation, et que les titres doivent être en grave. De même manière, **vous avez modifié la feuille de style CSS** pour mettre un élément en surbrillance. 

## Exercises et observations

- [ ] Balisez un toponyme (`placeName`), titre d'ouvrage (`title`), titre de fonction (`roleName`) ou organisation (`org`) en proximité d'un nom de personne.
- [ ] Changez la surbrilliance de l'un indépendemment de l'autre pour créer un contraste entre catégories. 

Notre éditeur XML offre trois présentations du même fichier XML.

1. `Visual` : la « version Word », pour les humains.
2. `Show tags` : un pseudo-XML qui ne montrent que les balises principales.
3. `Source` : le XML dans son état, pour les machines.

La version humaine est traduite depuis la version machine par la CSS, et vos interactions avec le texte sont aussi traduites par un intermédiare.  

- [ ] Retrouve un même élément que vous avez balisé dans les trois modes.
- [ ] Dans la mode `Source`, essayez de changer un `persName` en `placeName` et observez ce qui se passe.
- [ ] Observez le résultat dans la mode `Visual`.
- [ ] Sauvegardez le document et essayez de rater le code XML dans la mode `Source`.

Dans la mesure où vous avez modifié le code brut du document XML, vous êtes maintenant _codeur_ (ou _codeuse_).
