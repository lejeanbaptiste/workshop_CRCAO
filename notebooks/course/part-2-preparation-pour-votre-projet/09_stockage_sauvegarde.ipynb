# 9. Stockage, sauvegarde et versioning des fichiers

La structure ne vaut que si elle survit à l'ordinateur, à l'application et au premier mauvais clic. Traitez le corpus comme un ensemble de fichiers liés : source XML, traduction, métadonnées, entités, schéma, exports et notes.

## Une organisation simple

- Conservez une copie originale, en lecture seule, séparée des fichiers de travail.
- Donnez aux fichiers des noms stables, sans `nouveau projet final final`.
- Gardez les exports dans un dossier distinct et régénérable.
- Ne mettez jamais de clé API, de fichier privé ou de liste de contacts dans Git.

## Sauvegarder

Utilisez au moins deux supports ou services différents et vérifiez régulièrement qu'une restauration est possible. La synchronisation cloud n'est pas une sauvegarde si elle répercute immédiatement une suppression.

## Versionner avec GitHub

Git enregistre des états successifs. Faites des commits courts qui disent ce qui a changé : ajout de balises de lieux, correction d'une traduction, modification du schéma. Avant de publier, contrôlez les fichiers inclus et utilisez `.gitignore` pour les données privées.

Dans JupyterLite, le travail est conservé dans le stockage local du navigateur. Téléchargez les notebooks et les résultats si vous voulez les conserver ailleurs. Un autre navigateur ou un autre ordinateur ne possède pas automatiquement votre session.
