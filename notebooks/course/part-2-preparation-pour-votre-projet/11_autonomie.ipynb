# 11. Autonomie : questionnement et exécution

L'autonomie ne consiste pas à connaître **toutes** les balises TEI ni à écrire du code sans aide. Elle consiste à transformer **votre question de recherche** en une suite d'opérations **explicites**, **testables** et **documentées** — puis à itérer quand le texte résiste.

Vous avez déjà les briques : balisage (ch. 4–5), traduction séparée (ch. 3), désambiguïsation (ch. 7), contrôle des données (ch. 8), fichiers et sauvegardes (ch. 9), validation et schéma (ch. 10). Ce chapitre les **aligne** sur *votre* projet.

## De la question au modèle

Avant d'ouvrir l'éditeur, notez les réponses (même provisoires) à ces cinq questions :

1. **Quelle information** voulez-vous retrouver, compter ou comparer ?
2. **Quelle unité** compte : une mention, un passage, un paragraphe, un document, une entité ?
3. **Quelle balise TEI** (ou quelle structure) représente cette information ?
4. **Quels attributs** distinguent les cas (`@type`, `@key`, `@cert`, `@n`, …) ?
5. **Quel résultat concret** vous convaincrait que la question a reçu une réponse (table, graphique, export, liste) ?

Commencez **petit** : trois paragraphes, un chapitre, un auteur. Cherchez les exceptions, ajustez le modèle, puis élargissez. Une catégorie que vous ne savez pas interroger en XPath ou en pandas est peut-être une catégorie **mal définie** — ou inutile pour votre question.

## Exemple : transmissions maître–disciple d'un sutra

**Objectif :** tracer le réseau des transmissions entre maître et disciple pour un sutra donné, avec preuve textuelle et provenance.

**Entités à repérer :**

- personnes → `persName` (puis `@key` après désambiguïsation) ;
- titre du sutra → `title`.

**Relations à capturer :**

- qui transmet (`transmitteur`) ;
- qui reçoit (`récepteur`) ;
- quel texte ou objet est transmis.

**Métadonnées de provenance :**

- la **citation** ou formulation exacte dans la source ;
- le **titre** de l'ouvrage ou du manuscrit ;
- le **folio / numéro de page** si disponible.

### Demander conseil à une IA (sans lui confier le corpus)

Vous pouvez décrire le problème **sans** coller votre texte intégral (droits, confidentialité). Exemple de prompt :

```
Je prépare le balisage XML (TEI P5, schéma TEI All) d'un corpus sur les transmissions
maître–disciple d'un sutra. Je veux produire une table avec : le maître, le disciple,
le titre du sutra, la chaîne de caractères qui atteste la relation, le titre de la
source et le numéro de page.

Je compte utiliser persName pour les personnes et title pour le sutra.
Comment structurer le reste en TEI ? Faut-il des éléments dédiés, des attributs @role,
ou un segment englobant ? Proposez un extrait XML minimal valide.
```

Une réponse plausible (à **vérifier** dans les Guidelines et dans **Validation**) :

- `persName` avec `@role` (ou `@type`) pour distinguer maître et disciple — pas un élément `<person>` séparé dans le **texte** ; la personne réelle vit dans la base SQLite via `@key` ;
- un segment `<seg type="transmission">` (ou `<ab type="transmission">`) qui **encadre** les personnes, le titre et la formulation du témoignage ;
- `<pb n="…"/>` ou métadonnées de pagination selon votre modèle de source.

Exemple minimal (illustratif) :

```xml
<seg type="transmission">
  <persName role="master" key="person-000001">慧可</persName>
  transmet le <title>楞伽經</title> à
  <persName role="disciple" key="person-000002">僧璨</persName>.
</seg>
<pb n="12"/>
```

L'IA propose ; **vous** validez dans TEI, dans votre schéma (ch. 10), et sur **vos** trois paragraphes tests.

### Stratégie de balisage (ordre des opérations)

1. Baliser les **noms propres** et les **titres** (`persName`, `title`) — habitudes des ch. 4–5.
2. **Désambiguïser** les personnes (`@key`, ch. 7).
3. Encadrer les **passages de transmission** en `<seg type="transmission">` (souvent à la main ou par repérage ciblé, pas en regex aveugle sur tout le corpus).
4. Ajouter `@role` (ou conventions `@type`) de façon **cohérente** ; propager avec prudence.
5. Contrôler : **Validation**, tableau de fuites (ch. 8), petit export CSV quand les clés sont stables.

Si `@role="master"` n'est pas accepté en Visual, le schéma ou la définition de l'attribut doit être clarifiée (TEI standard vs personnalisation Roma, ch. 10).

## Documenter votre modèle

Une demi-page suffit souvent. Notez :

- la **question** ;
- les **balises et attributs** retenus (avec un exemple XML) ;
- ce qui est **hors scope** (« on ne balise pas encore les dates ») ;
- un **contre-exemple** volontairement non balisé (pour éviter la dérive).

Ce document sert à vous-même dans six mois, à un·e collègue, ou à une IA pour la suite du projet.

## Demander de l'aide (humaine ou outil)

Une bonne demande contient :

- le **but** en une phrase ;
- un **extrait XML minimal** (anonymisé si besoin) ;
- le **résultat attendu** vs **obtenu** ;
- les **étapes** pour reproduire le problème.

Pour l'éditeur LJB : [ouvrir une issue sur GitHub](https://github.com/lejeanbaptiste/lejeanbaptiste/issues) (bug ou idée). Pour votre corpus : pas de clés API ni de textes protégés dans le dépôt public.

## Et maintenant ?

Revenez à la question notée au ch. 0 ou au ch. 1. Vous avez un flux : lire, traduire, baliser, désambiguïser, contrôler, sauvegarder. L'autonomie, c'est d'enchaîner ces gestes **sans attendre** la prochaine séance — et de documenter assez pour ne pas vous perdre dans le bac à sable.

Un bug bien décrit devient une amélioration possible de l'outil. Une spécification floue devient des mois de balisage incohérent. Investissez d'abord dans la **clarté de la question**.
