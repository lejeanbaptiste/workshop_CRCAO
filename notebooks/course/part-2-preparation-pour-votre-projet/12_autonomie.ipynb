# 10. Autonomie : questionnement et exécution

L'autonomie ne consiste pas à connaître toutes les balises ni à écrire du code sans aide. Elle consiste à transformer une question en une suite d'opérations explicites, testables et documentées.

## De la question au modèle

1. Quelle information voulez-vous retrouver ou comparer ?
2. Quelle unité compte : une mention, un paragraphe, un document, une personne ?
3. Quelle balise représente cette information ?
4. Quels attributs sont nécessaires pour distinguer les cas ?
5. Quel résultat vous convaincrait que la question a reçu une réponse ?

Commencez petit. Testez votre modèle sur trois paragraphes, cherchez les exceptions, puis élargissez. Une catégorie dont vous ne savez pas quoi faire dans une requête est peut-être une catégorie décorative.

## Modifier le schéma

Si votre projet demande une structure que le schéma ne permet pas, documentez d'abord le besoin avec des exemples. Décrivez l'élément, ses parents possibles, ses attributs, ses répétitions et un cas qui doit être refusé. Ce document sera plus utile qu'une modification improvisée.

## Demander de l'aide

Une bonne question technique contient : le but, un extrait minimal, le résultat attendu, le résultat obtenu et l'étape qui reproduit le problème. Pour l'IA, ajoutez le contexte et les contraintes ; ne transmettez pas de données confidentielles. Pour GitHub, créez une Issue claire.

## Exercice final

Écrivez une mini-spécification de votre prochaine analyse : question, balises nécessaires, attributs, échantillon de test, sortie attendue et limites. Vous avez alors un plan de travail transmissible à une autre personne — ou à vous-même dans six mois.
