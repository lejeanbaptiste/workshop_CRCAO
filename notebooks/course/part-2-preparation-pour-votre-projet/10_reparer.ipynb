# 11. Je l'ai cassé, que faire ?

Tout le monde casse du XML. L'objectif n'est pas de ne jamais faire d'erreur : c'est de rendre l'erreur localisable et réversible.

## Si le XML est invalide

- Lisez le premier message de `Validation`, pas seulement le dernier.
- Revenez à la dernière modification.
- Vérifiez les balises ouvertes et fermées, leur imbrication et les guillemets des attributs.
- Comparez avec une copie valide ou utilisez `Source` pour repérer le caractère fautif.

## Si les liens ne fonctionnent plus

Les traductions, les entités et les documents XML peuvent être liés par des identifiants, des chemins ou des clés. Ne renommez pas et ne déplacez pas les fichiers au hasard. Vérifiez le projet, les chemins relatifs, les identifiants `key` et la présence des fichiers associés.

## Si une automatisation a mal tourné

Arrêtez-vous avant de propager la correction. Travaillez sur une copie, filtrez les suggestions, rejetez la série et recommencez avec un échantillon plus petit. Une opération rapide est utile seulement si elle est contrôlable.

## La procédure de récupération

1. Ne supprimez pas l'original.
2. Notez ce qui a été fait et quand.
3. Identifiez le dernier état valide.
4. Restaurez ou comparez avec la sauvegarde.
5. Reproduisez le problème sur un extrait minimal.
6. Demandez de l'aide avec cet extrait, le message d'erreur et les étapes.
