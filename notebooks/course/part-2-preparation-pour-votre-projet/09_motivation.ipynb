# 9. Stockage, sauvegarde et partage des fichiers

Vous avez montré jusqu'ici que vous savez suivre des instructions, mais il faut vous préparer à l'**autonomie**, car le balisage XML est un bac à sable. Commencons alors par la question des fichiers.

## L'organisation des fichiers locaux

Votre éditeur trait deux types de fichiers : les fichiers utilisateur (source primaires, translations et base de données) et les ressources, configurations, etc. L'éditeur vous permet de modifier ce que vous êtes censé modifier, mais vous êtes effectivement libre de trouver votre base de données et de la corrumpre. C'est un risque réel lorsque vous lancez des scripts Python pour traiter l'ensemble d'un corpus. 

L'important :

- Votre base de données (`entities.xml`), ressources téléchargés et fichiers de config vivent dans le dossier racine que vous avez sélectionné dans le panneau de configuration.
- Le plus simple est de stocker tous vos dossiers de projet comme sous-dossiers du dossier racine.
- Chaque dossier de projet contient trois dossiers :
    - `.ljb-time-machine` (caché) : vos sauvegardes
    - `.ljb` (caché) : vos configurations
    - `schema` : le schéma XML, la CSS, etc.
- Vos fichiers sont maintenant liés les uns aux autres : les traductions sont liées aux paragraphes dans un autre fichier par les UUID des paragraphes ; les identifiants des personnes, des lieux, etc., sont liés à votre base de données. C'est pour cette raison que _tous vos fichiers doivent voyager ensemble_.

## Sauvegarde

Vous allez à un moment faire une _grosse_ bêtise, comme remplacer un nom de personne par un autre dans un corpus de 10 000 textes. Les fichiers XML sont très petits, donc il est encouragé de faire régulièrement un snapshot « Time Machine ». Après votre bêtise, vous pouvez restorer l'ensemble de vos fichiers.

## Synchronisation par service cloud

Synchroniser vos fichiers dans le cloud est une bonne idée. Assurez-vous simplement de _tout_ synchroniser, notamment le dossier racine. Il faut que vos corpus voyage avec la base de donnés qui va avec.

## Github

Un autre possibilité à explorer est GitHub et son [appli graphique](https://github.com/apps/desktop), si la ligne de commande vous fait peur. Mais c'est un autre paradigme :

- Créez un noueau repo dans GitHub. Choissisez privé, s'il est juste pour toi ou contiendra des fichiers que vous n'êtes pas censer diffuser.
- Associez-le à votre dossier de projet dans l'appli
- Ajouter une liste de fichiers à ignorer (`.gitignore`), notamment vos configs et sauvegards dans les dossiers cachés.
- « Commit » et « push » vos fichiers locaux vers GitHub.

La prochaine fois que vous voulez sauvegarder, « commit » et « push ». Pour télécharger la plus récénte version de vos fichiers (en cas de collaboration ou du travail sur un autre ordi), « pull ».

GitHub is of potential interest to you for the following reasons. First, all one needs to do to publish one's corpus is change it from private to public. Second, GitHub stores every version of your corpus that you send it, and it will show you the differences line-by-line. That way, if you are collaborating with someone, you can see every single thing that he changed and, if it causes a problem, return to the previous version. The disadvantage is that if there is a _conflict_, you have to manually sort it out, and that can be a little confusing.
