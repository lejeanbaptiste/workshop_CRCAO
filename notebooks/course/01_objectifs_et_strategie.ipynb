# 1. Objectifs et stratégie

Vous êtes ici parce que votre projet implique déjà, directement ou indirectement, le XML. Je connais personnellement votre projet, vous êtes motivé pour l'avancer, et j'ai quelques compétence qui peuvent potentiellement vous aider. J'ai conçu alors cette séance et l'outil principal en fonction de vos besoins individuels – chacun d'entre vous. 

Nous partons d'**un flux de travail familier** : lire une source, la traduire, prendre des notes, repérer des motifs. L'obectif n'est ni de le transformer ni de le remplacer, mais de le modifier légèrement pour pouvoir faire « une pierre deux coups » dans l'intérêt de **l'efficacité**.

Le travail de balisage XML implique souvent un investissment lourd en amont pour un retour nébuleux, à long terme. Pour vous motiver, nous allons donner quelques exemples de ce que _ça peut donner_ afin de créer des **récompenses concrètes et immédiates**. 

Finalement, **nous mettrons la pratique avant la théorie**: on essaie, on observe, on maîtrise une geste, puis on lui donne un nom et une explication.

« Pourquoi tu fais tout cela pour nous, Daniel ? » Mais non, vous n'avez rien compris : j'ai promis une formation en XML afin de récruter une équipe de bêta-testeurs – sur de différents systèmes d'exploitation et architectures de puce – qui va œuvrer gratuitement, sous ma surveillance, pendant toute une journée. À chaque fois que vous rencontrer un problème, vous allez [déposer une « Issue » sur GitHub](https://github.com/lejeanbaptiste/lejeanbaptiste/issues). Si vous apprenez quelque chose, tant mieux.
