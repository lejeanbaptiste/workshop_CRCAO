# 3. Préparation et traduction d'une source primaire « à l'ancienne »

## Pratique traditionnelle

Le travail d'un sinologue consiste avant tout à lire et à traduire des sources primaires. Voici notre flux de travail traditionnel, quasiment inchangé depuis l'époque des Qing :

- Allumer une bougie.
- Faire une prière aux ancêtres. 
- Copier-coller une transcription trouvée en ligne dans Word.
- Consulter [Zdic](https://zdic.net/) et [Le Grand Ricci Online](https://brill.com/display/db/lgr) pour les mots difficiles, ou même un PDF de 故訓匯纂 s'il faut vraiment aller fouiller dans des livres.
- Mettre en surbrillance des passages important sans aucune explication de pourquoi l'on a fait ça – on s'en souviendra un jour, et ça sera génial. 
- Taper une traduction et des notes après le paragraphe étudié.
- Sauvegarder le fichier sous le nom `nouveau projet.docx` dans le dossier `sans titre` quelque part avec nos déclarations d'impôts.

## Petite modification

- Ouvrez l'éditeur XML.
- Ouvrez un nouveau projet.
- Choissez le schéma « TEI ALL » (voir XX)
- Dans `Project settings`:
	- entrez la langue de vos sources primaires
	- selectionnez la langue de vos traductions
	- laissez les défauts (central database, Paragraph (1:1))
- Importation :
  - soit
    - créez un nouveau document,
    - collez votre transcription dans « paragraph text » ;
  - soit
    - importez un document (`Menu>File>Import`).
- Ouvrez le panneau « Translation » à droite.
- Cliquez sur le paragraphe qui vous intéresse et taper une traduction dans le panneau « Translation ».

Dans le panneau « Translation », vous pouvez ajouter des notes de bas de page, insérer des références Zotero, etc. Cette traduction est liée au paragraphe selectionné depuis un fichier séparé, pour que l'on ne mélange pas la source primaire et vos traductions. Ainsi, vous pouvez aussi attacher vos traductions en différentes langues à une seule et unique transcription, les cacher et exporter votre transcription avec la traduction de votre choix.

Si vous avez fourni une clé API, vous pouvez aussi auto-générer une traduction. Soyons clairs : la qualité dépend du modèle, et les APIs gratuites n'ont pas les meilleurs modèles, mais on peut toujours s'y appuyer pour la lecture rapide des sources difficiles et en tirer de l'inspiration.

Finalement, exportez votre document en format Word avec (ou sans) la traduction de votre choix.

Ce n'est pas loin de votre flux de travail actuel. Celui-ci présente des avantages (unification des transcriptions, séparation des traductions) ansi que des inconvenients (toutes les options de mise en forme). Mais je suis le dévelopeur, et ce n'est que la version bêta.

## Exercices

**Gestes de base**

- [ ] Créez un document
- [ ] Ouvrez et fermez des documents
- [ ] Copiez-collez un texte dedans.
- [ ] Importez l'un de vos propres sources primaires.
- [ ] Parcourez les document dans le panneau « Navigator »
- [ ] Entrez le titre et d'autre informations sur le document dans le panel « Document metadata »
- [ ] Ajoutez quelques notes sur plusieurs paragraphes et les retrouver
- [ ] Ouvrez « Find and replace » (panel gauche, `ctrl/cmd+f`) pour chercher des mots clés dans la source primaire et dans vos traductions.

**Intégration**

- [ ] Traduisez un petit paragraphe de l'une de vos sources en ajoutant des notes de bas de page et quelques références Zotero/Juris-M.

## Point de vigilance

Ne paniquez pas : rien n'est perdu, vos traductions sont là, il faut juste ouvrir le panneau qui y est dédié.
