### 3. Regex

Dans l'exemple précedent, j'ai écrit `<persName>NOM</persName>` est vous avez compris que l'on n'écrit pas « NOM » mais plutôt la chaîne de caractères qui vous intéresse, n'est-ce pas ? La chaîne « NOM » dans cet exemple est un **variable**, est il est possible de communiquer des variables de manière très précise à la machine. Ce langage de variables textuelles s'appele _regex_ ([regular expressions](https://fr.wikipedia.org/wiki/Expression_r%C3%A9guli%C3%A8re), ou « expressions régulières »).

Voici quatre simples _regex_ à apprendre :

- `\w` : une « lettre » de mot (**w**ord) quelconque
- `\n` : nouvelle ligne
- `+` : « ou plusieurs »
- `()`: « capturer pour le reproduire dans le résultat »
- `\1`: « reproduire la première chaîne capturée »

Les _regex_ sont très utiles. Prenons pour exemple le passage suivant : 

**《太平御覽·夏下》：**

>《易說》曰：立夏清明風至而暑，鶴鳴博穀飛，電見龍升天。〈龍，心星名。〉
>《易通卦驗》曰：立夏雨，螻蛄鳴。
>《三禮義宗》曰：四月立夏爲節者，夏，大也，至此之時，物已長大，故以爲名。小滿爲中者，物之生長小得幷滿，故以小滿爲名也。
>《孝經緯》曰：穀雨後十五日，鬥指辰東南維爲立夏，後十五日，鬥指巳爲小滿。
>《續漢書·禮儀志》曰：立夏之日，夜漏未盡五刻，京師百官皆衣赤，至季夏衣黃。
>《抱樸子》曰：或問不熱之道。答曰：「立夏之日，或服玄冰丸，或服飛霜散及六壬六癸之符，則不熱。幼伯子、王仲都，此二人衣之以重裘，曝之于夏日之中，周以十爐之火，口不稱熱，身不流汗，蓋用此方者也。」

### 5. IA

Vous avez sans doute l'experience que l'IA est dangereux dans la mesure où elle halucine et risque de récrire vos sources primaires. Elle n'est pas dangereuse, elle a besoin simplement de l'encadrement. En bref :

- on donne une copie du texte à l'IA et on lui demande de remplir une formulaire avec les coordonnées de chaque modification qu'il veut faire dessus, 
- on demande la validation de l'utilisateur, 
- puis on met en place les corrections validées de façon méchanique, indépendemment de l'IA, en ajoutant seulement des balises. 

Pour ce faire, il faut que l'éditeur parle directement à l'IA par API ([Application Programming Interface](https://fr.wikipedia.org/wiki/Interface_de_programmation)). Quelle IA ? C'est à vous de choisir : vous pouvez en faire tourner une en local avec [Ollama](https://ollama.com/) ; vous pouvez utiliser l'API gratuite que proposent quelques services comme [Groq](https://console.groq.com/) ; sinon, vous pouvez payer pour une clé chez [Mistral](https://chat.mistral.ai/), [ChatGPT](https://chatgpt.com/), etc.
