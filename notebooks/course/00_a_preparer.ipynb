````python
import ipywidgets as widgets
from IPython.display import display


def render_checklist(sections):
    blocks = []
    for title, items in sections:
        blocks.append(widgets.HTML(f"<h4 style='margin:0.6em 0 0.3em 0'>{title}</h4>"))
        for item in items:
            blocks.append(widgets.Checkbox(value=False, description=item, indent=False))
    display(widgets.VBox(blocks))
````
