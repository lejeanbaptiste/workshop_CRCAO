# 0. À préparer en amont

## Matériels

- [ ] Ordinateur portable
- [ ] Smartphone

## Expertise préalable

- [ ] À l'aise dans un traitement de texte, comme Word. 
- [ ] Téléchargement et installation des logiciels.
- [ ] Ouvrature des comptes en ligne.
- [ ] Gestion de mots de passe. 

Si vous avez créé un compte, mais vous avez oublié le mot de passe, ou vous l'avez noté de fait dans votre autre ordinateur qui est à la maison... Je vous mettrai un châpeau d'âne et vous demanderai de créer un nouveau compte pendant que l'on s'amuse.  

## Textes

Ce workshop part de votre projet, pas d'un exemple abstrait. Venez avec : 

- [ ] 2 à 5 transcriptions du type de source que vous souhaitez étudier, 
- [ ] dont au moins une contenant des noms de personnes, de lieux, d'institutions ou de titres d'ouvrage.

## Questions

- [ ] Notez quelques questions/objectifs de recherche concrets : que voudriez-vous regrouper, compter, comparer ou retrouver ? Que voulez-vous fabriquer ? _Pas de l'orde de la magie téchnologique mais de la **recherche scientifique**_.

Par ex. :

- « Je veux voir s'il n'y a pas une signification quelconque derière les listes de textes reçus dans des biographies tibétains »
- « J'aimerais bien faire une base de données avec toutes les citations de Confucius dans les manuscrits des Royaumes Combatants, ou l'on peut trier par... »
- « Je souhaite extraire des formules dispersés "_A_ est maître de _B_," "_C_ est fils de _D_," un réseau étendu de relations.  »

## Listes

- [ ] _Si vous en avez..._ toute liste de noms ou de termes d'une catégorie précises, par exemple la liste de 1000 noms de personages en Pinyin et chinois avec dates de naissance et de déces que vous gardez comme référence dans vos notes. 

## Comptes

Vous allez créer un **compte [GitHub](https://github.com/)** dans l'objectif simple (1) d'avoir une présence numérique dans cet espace, (2) d'explorer une nouvelle forme de stockage, de partage et de collaboration et (3) d'activer une fonctionalité importante de l'éditeur XML, à expliquer plus tard. 

- [ ] Créez un compte [ici](https://github.com/)
- [ ] Télécharger l'appli GitHub sur votre téléphone portable.
- [ ] [Activer l'authentification à 2 facteurs](https://docs.github.com/fr/authentication/securing-your-account-with-two-factor-authentication-2fa/configuring-two-factor-authentication) afin de valider rapidement des connexions depuis votre portable.

Vous allez aussi créez un compte developeur gratuit chez [Groq](https://console.groq.com/) (≠ Groc), qui offre une **clé API à des LLM**, c'est à dire un canal pour que votre ordinateur puisse echanger directement avec une IA qui tourne sur leur serveur. 

1. Créez un compte
2. Ouvrez le truc dans le coin droit supérieur de la page d'accueil
3. Créez une clé API
4. Notez la quelque part

## Logiciels

Vous aurez besoin d'un **éditeur de texte brut avec regex**:

- Linux : gedit (`sudo apt-get install gedit')
- MacOS : [cotEditor](https://coteditor.com/)
- Windows : [Notepad++](https://notepad-plus-plus.org/)

Vous aurez besoin d'un **éditeur XML** avec interface graphique (« pour les nuls »). Il n'y a rien de parfait, si vous en cherchez dans Google, mais il existe de petits projects sur GitHub qui sont parfois supérieurs aux solutions commerciales.

- [ ] Allez dans [ce repo](https://github.com/lejeanbaptiste/lejeanbaptiste), qui va être complètement incompréhensible.
- [ ] Pour avoir l'expérience, lire la description pour vérifier qu'il s'agit bien d'un éditeur XML (c'est le cas, ne vous inquiétez pas).
- [ ] Cliquez sur l'etiquette « v0.0... » en-dessous de « Releases » pour vous diriger vers les installateurs. 
- [ ] Téléchargez le paquet qui corréspond à votre système d'exploitation (Linux/macOS/Windows) et à votre puce (AMD/ARM).[^1]
- [ ] Installer le paquet

Une fois l'éditeur installé, ouvrez-le et paramétrez-le :

- [ ] Votre nom (pour l'encoder dans les méta-données des textes que vous éditez)
- [ ] Dosser racine, ou il stockera votre base de donnés et les ressources téléchargése.
- [ ] API IA :  

```
Base URL: https://api.groq.com/openai/v1
API key: [votre clé API, voir ci-dessus]
Model: qwen/qwen3.6-27b
```

Au cas ou cet éditeur bogue ou réfuse de s'installer, vous pouvez installer [XMLmind](https://www.xmlmind.com/products_fr.html) comme plan B.
