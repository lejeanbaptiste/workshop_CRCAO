# 7. Désambiguïsation

## Pratique traditionnelle

Lorsque l’on travaille sur une source primaire, on ne s’arrête pas, bien sûr, à la catégorisation des mots sur la page — c’est-à-dire à la question : « Est-ce une personne, cela ? ». On cherche plutôt à établir un lien entre les mots et les réalités qu’ils désignent.

Pour un historien des Han, les mots « Liu Xin » 劉歆 renvoient à une personne : un lettré célèbre, bibliothécaire et propagandiste, qui a accompli des actes, rédigé des ouvrages, entretenu des liens familiaux, sociaux, politiques et académiques, et dont la vie s’inscrit dans des dates précises de naissance, de décès et d’activité. Ce sont précisément ce genre de réalités qui nous intéressent en histoire : les personnes, les lieux, les ouvrages, les institutions, ainsi que les titres, les fonctions et les dates.

Mais de quoi s’agit-il au juste ? Ce ne sont pas des objets que l’on peut voir, toucher ou consulter directement… Ce sont des « entités », c’est-à-dire des constructions conceptuelles fondées sur une collection de données externes. Au fond, « Le soi ne se connaît pas de façon immédiate, mais par le détour de repères culturels, parmi lesquels figurent au premier chef les récits » (Paul Ricœur, « L'identité narrative »).

**Qui est Liu Xin ?** Si vous êtes comme moi, vous n’avez pas toujours toutes les informations sur une telle entité en tête.
Vous avez pris des notes quelque part. Avant de rédiger un paragraphe à son sujet, vous consultez Wikipédia et des ouvrages de référence, notamment pour vérifier ses dates de vie.

Il y a une autre étape, si naturelle qu’on n’y pense plus : Lorsque vous tapez « Liu Xin » dans Google, le moteur ne vous propose qu’une présentatrice et une rappeuse. Vous essayez ensuite sur Wikipédia, qui vous redirige vers une liste de toutes les personnalités célèbres portant ce nom. C’est ce qu’on appelle une « page de désambiguïsation » (disambiguation page) :

<img src="../images/wikipedia-disambiguation-liu-xin.png" width="500">

Comment procéder pour « désambiguïser » ? Après chaque « Liu Xin », Wikipédia ajoute une description courte (« scholar ») ou détaillée (« 劉歆 ; vers 50 av. J.-C. – 23 ap. J.-C., astronome, historien, bibliothécaire… »). Vous cliquez sur la bonne entrée, et la page dédiée vous fournit ses dates, son nom de style, sa parenté, etc.

Vous faites tout cela sans y réfléchir. Je vais vous proposer une approche plus efficace, puis vous inviter à analyser vos pratiques pour les adapter à votre lectorat informatique.

## Petite modification

Après avoir balisé votre texte, ouvrez le panneau **`Disambiguate`**. Pour chaque chaîne balisée, LJB propose des **entités** (personne, lieu, ouvrage…) trouvées dans les autorités activées dans la configuration du projet — par exemple le [Buddhist Studies Authority Database Project](https://authority.dila.edu.tw/). Comme sur Wikipédia, vous cliquez sur le bon « Liu Xin » et vous associez toutes les occurrences de ce nom dans votre texte à **cette seule personne**. 

<img src="../images/08_disambiguate.png" width="1000">

En acceptant, le choix est enregistré dans votre projet :

- la personne entre dans **votre** base de données locale SQLite ;
- elle reçoit un **identifiant unique** ;
- cet identifiant est relié de façon durable à celui de l'autorité — et donc aux informations de la base externe ;
- le lien vers Wikipédia (ou une autre autorité) apparaît dans le panneau **`Attributes`** ;
- vous retrouvez la fiche dans **`Entities`** : vous pouvez y ajouter des noms alternatifs, corriger une association, etc.

<img src="../images/09_database_card.png" width="1000">

Parfois plusieurs autorités décrivent la même personne : en choisissant la bonne entrée Wikipédia, vous récupérez aussi la concordance avec CBDB, DILA, VIAF ou la NDL. Parfois il y a des doublons à regrouper. Parfois il n'y a rien : vous pouvez quand même créer une **nouvelle entité** locale.

**Cas particulier :** le prénom 歆, tout seul. Il n'est souvent ni balisé automatiquement ni reconnu tel quel par `Disambiguate`. Ce n'est pas grave :

1. Sélectionnez une occurrence.
2. Balisez-la en `persName`.
3. Ouvrez **`Attributes`**, cherchez le nom complet dans **`Lookup`**, associez l'entité.
4. Propagez le balisage et l'association aux autres occurrences.
5. Ajoutez « 歆 » comme forme de surface de 劉歆 dans **`Entities`**.

## Usages immédiats

### 1. Un Zotero pour le reste

<img src="../images/11_database_viewer.png" width="1000">

Une fois qu’une balise est désambiguïsée, il n’est plus nécessaire de recourir à Google pour retrouver les informations sur une personne, un ouvrage, un toponyme, etc. : toutes les données de base et les liens d’autorité sont centralisés dans votre propre base de données, consultable hors ligne. Le **Database viewer** vous permet en outre d’ajouter des notes à chaque entrée, à l’instar de Zotero.

### 2. Extension Word

Lorsque l'on écrit un article, la saisie et la mise en forme des noms, sinogrammes, traductions et dates pour chaque entité peuvent s’avérer aussi fastidieuses que l’étaient autrefois les bibliographies. Pour remédier à cela, j’ai développé une extension pour Word, LibreOffice et OnlyOffice qui automatise ce processus à partir de votre base de données, « à la Zotero ».

<img src="../images/12_word_plugin.png" width="1000">

### 3. Traduction IA améliorée

Les outils d’IA traduisent de mieux en mieux nos textes historiques. En revanche, ils traitent souvent de manière superficielle, voire trompeuse, les données qui nous sont les plus précieuses : l’identification, la compréhension, la traduction exacte et la mise en forme des dates, des noms propres, des titres, etc. Une fois le texte balisé, LJB remplace les entités par des variables dans la version soumise à l’IA. Celle-ci traduit le texte autour de ces variables, puis LJB réinsère les entités sous forme de champs, comme dans Zotero. Ainsi, c’est vous qui définissez la traduction et la mise en forme à appliquer, d’un seul coup, à l’ensemble de votre corpus.

En chinois et japonais, l'IA ne se trompe _jamais_ sur **les dates**, car les dates sont gérées par une instantanée de [Sanmiao](https://norbert.huma-num.fr/app/sanmiao/index.html?lang=fr), un convertisseur de dates qui exige la précision scientifique pour l'histoire de l'astronomie.

Également, le paquet sinologique exploite les 8900 traductions dans Hucker (_Dictionary Of Official Titles In Imperial China_) pour ce qui concerne les titres officiels. LJB n'a évidemment pas le droit de _distribuer_ ces traductions, donc il mine, sur votre ordinateur, celles qui ont été incluses dans la version téléchargeable du CBDB. Pour aller plus loin, j'ai distillé ce dictionnaire en algorithmes et entraîné une IA à traduire « à la Hucker » pour étendre ses traductions à plus de 22,000 titres, en anglais (HuckBot5000) et en français (MaxiRicci7000). Ainsi, une fois balisés et désambiguïsés, les titres seront toujours traduits de manière fiable et homogène par l'IA.

## Désambiguïsation : théorie informatique minimale

Dans le XML, la désambiguïsation consiste à relier une balise à un entité précise à l'aide d'un **identifiant** (`key`).

Pour gérer des identifiants de manière fiable, il faut une infrastructure qui dépasse normalement les capacités d’un chercheur isolé : attribution d’identifiants uniques, regroupement des formes de surface, concordances entre bases de données. C’est pourquoi existent les **« autorités »** — Wikipédia, CBDB, DILA, VIAF, la NDL, etc.

Cependant, ces autorités utilisent **leurs propres systèmes d’identifiants**. Elles se recoupent, se renvoient les unes aux autres, et laissent inévitablement des lacunes. Surtout, **vous ne pouvez pas les corriger ou les compléter** selon les besoins spécifiques de **votre projet** : vous n’allez pas réécrire Wikipédia pour chaque personnage secondaire de vos sources, ni fusionner manuellement toutes les bases de données du monde.

Il vous faut donc **votre propre base de données relationnelle**, minimale mais sous votre contrôle. Quoi faire ? LJB vous en crée une pour vous à l'installation qui aspire automatiquement les données dont vous avez besoin depuis les autorités de votre choix.

## Attributs

On distingue les **chaînes de caractères** et les **entités**. L'identifiant est inscrit dans la balise sous forme d'**attribut** :

```xml
<persName key="person-000020" resp="#ljb-autotag">許芝</persName>
```

Ici, `key` pointe vers votre base locale ; `resp` indique l'origine du balisage. En mode `Visual`, le panneau **`Attributes`** affiche la même information. Vous pouvez aussi ajouter `cert` pour indiquer votre degré de certitude. Voir la [feuille de style TEI pour `persName`](https://www.tei-c.org/release/doc/tei-p5-doc/fr/html/ref-persName.html) pour d'autres attributs (`@type`, etc.).

Comme au chapitre 6, XPath sert aussi à **contrôler** le balisage :

Toutes les balises avec un attribut `cert` :

```xpath
TEI//*[@cert]
```

Toutes les balises avec `cert="5"` :

```xpath
TEI//*[@cert="5"]
```

Tous les `persName` sans identifiant :

```xpath
TEI//persName[not(@key)]
```

Toutes les mentions de la personne `person-000020` :

```xpath
TEI//persName[@key="person-000020"]
```

- [ ] Essayez l'outil `Disambiguate` sur votre texte balisé
- [ ] Choisissez la bonne entité pour un homonyme (comme sur Wikipédia)
- [ ] Appliquez le choix à toutes les occurrences similaires
- [ ] Vérifiez le lien vers l'autorité dans `Attributes`
- [ ] Trouvez un prénom solitaire ou un nom non balisé
- [ ] Désambiguïsez-le via `Lookup` dans `Attributes`
- [ ] Propagez le balisage et l'association dans le document
- [ ] Ouvrez `Attributes` sur une balise déjà désambiguïsée
- [ ] Ajoutez `cert="5"` à deux balises dont vous êtes sûr
- [ ] Consultez `Entities` : modifiez une fiche, ajoutez une forme de surface
- [ ] Testez `TEI//persName[not(@key)]`
- [ ] Testez `TEI//*[@cert="5"]`
- [ ] Testez `TEI//persName[@key="..."]` pour une entité que vous venez de lier
