# 7. Désambiguïsation

## Pratique traditionnelle

Vous avez oublié les dates de 劉歆, le fameux lettré, bibliothécaire et propagandiste à la fin des Han orientaux. Vous tapez « Liu Xin » dans Google, mais il ne vous propose qu'une présentatrice et une rappeuse. Vous le cherchez ensuite dans Wikipédia, qui vous dirige vers une liste de tous les personnages fameux qui portent le même nom. Ceci s'appelle une « page de désambiguïsation » (_disambiguation page_) :

![Page de désambiguïsation Wikipédia pour « Liu Xin »](../images/wikipedia-disambiguation-liu-xin.png)

Comment désambiguïser ? Après chaque « Liu Xin », Wikipédia ajoute une description courte (« scholar ») ou longue (« 劉歆; c. 50 BCE – 23 CE), astronomer, historian, librarian… »). Vous cliquez sur la bonne entrée ; la page du bon Liu Xin vous donne ses dates, son nom de style, sa parenté, etc.

Vous savez déjà faire cela. C'est une habitude de recherche.

## Petite modification

Dans LJB, c'est **la même habitude**, mais intégrée à votre traitement de texte : Wikipédia, CBDB, DILA, VIAF, la NDL et d'autres **autorités** sont accessibles depuis l'éditeur, sans changer de fenêtre.

Après avoir balisé votre texte, ouvrez le panneau **`Disambiguate`**. Pour chaque chaîne balisée, l'éditeur propose des **entités** (personne, lieu, ouvrage…) trouvées dans les autorités activées dans la configuration du projet — par exemple le [Buddhist Studies Authority Database Project](https://authority.dila.edu.tw/).

Comme sur Wikipédia, vous cliquez sur le bon « Liu Xin » et vous associez toutes les occurrences de ce nom dans votre texte à **cette seule personne**. La différence, c'est que le choix est enregistré dans votre projet :

- la personne entre dans **votre** base de données locale (`entities.xml`) ;
- elle reçoit un **identifiant unique** ;
- cet identifiant est relié de façon durable à celui de l'autorité — et donc aux informations de la base externe ;
- le lien vers Wikipédia (ou une autre autorité) apparaît dans le panneau **`Attributes`** ;
- vous retrouvez la fiche dans **`Entities`** : vous pouvez y ajouter des noms alternatifs, corriger une association, etc.

Parfois plusieurs autorités décrivent la même personne : en choisissant la bonne entrée Wikipédia, vous récupérez aussi la concordance avec CBDB, DILA, VIAF ou la NDL. Parfois il y a des doublons à regrouper. Parfois il n'y a rien : vous pouvez quand même créer une **nouvelle entité** locale.

**Cas particulier :** le prénom 歆, tout seul. Il n'est souvent ni balisé automatiquement ni reconnu tel quel par `Disambiguate`. Ce n'est pas grave :

1. Sélectionnez une occurrence.
2. Balisez-la en `persName`.
3. Ouvrez **`Attributes`**, cherchez le nom complet dans **`Lookup`**, associez l'entité.
4. Propagez le balisage et l'association aux autres occurrences.
5. Ajoutez « 歆 » comme forme de surface de 劉歆 dans **`Entities`**.

## Théorie minimale

Dans le XML, il n'y a pas de « personne », seulement des **mots** (`persName`, `placeName`…) qui renvoient à des choses du monde — des **entités**. La désambiguïsation consiste à relier chaque forme à **une** entité précise, puis à la substituer par un **identifiant stable** (`key`), pour ne plus mélanger des homonymes dans vos statistiques.

Pour tenir des identifiants de façon fiable, il faut une **infrastructure** qui dépasse ce qu'un chercheur seul peut construire et maintenir : attribution d'identifiants uniques, regroupement des formes de surface, concordances entre bases. C'est pour cela qu'existent les **autorités** — Wikipédia, CBDB, DILA, VIAF, la NDL, etc.

Mais ces autorités utilisent chacune **leurs propres identifiants**. Elles se recouvrent, se renvoient l'une à l'autre, et laissent forcément des trous. Surtout, **vous** ne pouvez pas les corriger ni les compléter au gré de **votre** projet : vous n'allez pas réécrire Wikipédia pour chaque personnage mineur de vos sources, ni fusionner à la main toutes les bases du monde.

Il vous faut donc **votre propre base de données relationnelle**, minimale mais sous votre contrôle. LJB vous en fournit une dans `entities.xml` : identifiants locaux, formes de surface regroupées, concordances vers les autorités auxquelles l'éditeur est connecté — et la possibilité d'ajouter ce que les autorités ignorent. Ce n'est pas un remplacement des autorités ; c'est le **minimum** qui fonctionne **avec** elles, pour que votre corpus et vos questions de recherche restent les vôtres.

## Attributs

On distingue les **chaînes de caractères** et les **entités**. L'identifiant est inscrit dans la balise sous forme d'**attribut** :

```xml
<persName key="person-000020" resp="#ljb-autotag">許芝</persName>
```

Ici, `key` pointe vers votre base locale ; `resp` indique l'origine du balisage. En mode `Visual`, le panneau **`Attributes`** affiche la même information. Vous pouvez aussi ajouter `cert` pour indiquer votre degré de certitude. Voir la [feuille de style TEI pour `persName`](https://www.tei-c.org/release/doc/tei-p5-doc/fr/html/ref-persName.html) pour d'autres attributs (`@type`, etc.).

Comme au chapitre 6, XPath sert aussi à **contrôler** le balisage :

Toutes les balises avec un attribut `cert` :

```xpath
TEI//*[@cert]
```

Toutes les balises avec `cert="5"` :

```xpath
TEI//*[@cert="5"]
```

Tous les `persName` sans identifiant :

```xpath
TEI//persName[not(@key)]
```

Toutes les mentions de la personne `person-000020` :

```xpath
TEI//persName[@key="person-000020"]
```

- [ ] Essayez l'outil `Disambiguate` sur votre texte balisé
- [ ] Choisissez la bonne entité pour un homonyme (comme sur Wikipédia)
- [ ] Appliquez le choix à toutes les occurrences similaires
- [ ] Vérifiez le lien vers l'autorité dans `Attributes`
- [ ] Trouvez un prénom solitaire ou un nom non balisé
- [ ] Désambiguïsez-le via `Lookup` dans `Attributes`
- [ ] Propagez le balisage et l'association dans le document
- [ ] Ouvrez `Attributes` sur une balise déjà désambiguïsée
- [ ] Ajoutez `cert="5"` à deux balises dont vous êtes sûr
- [ ] Consultez `Entities` : modifiez une fiche, ajoutez une forme de surface
- [ ] Testez `TEI//persName[not(@key)]`
- [ ] Testez `TEI//*[@cert="5"]`
- [ ] Testez `TEI//persName[@key="..."]` pour une entité que vous venez de lier

## Exercices

La désambiguïsation est une interprétation : elle peut être révisée. Un identifiant stable rend cette révision explicite. Au chapitre suivant, nous exploiterons des **entités** identifiées, pas seulement des chaînes.
